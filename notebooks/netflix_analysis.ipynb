# Netflix Movies and TV Shows Exploratory Data Analysis

**Dataset**: Kaggle / Shivam Bansal Netflix Shows Catalog (`netflix_cleaned.csv`)
**Author**: Senior Data Analyst
**Objective**: In-depth analysis of catalog distribution, content composition, temporal patterns, and international content diversification.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Set plotting style
plt.style.use('dark_background')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

# Load processed data
df = pd.read_csv('../data/processed/netflix_cleaned.csv')
print(f"Loaded {len(df):,} cleaned catalog entries with {len(df.columns)} features.")
df.head(3)
```

## 1. Content Type Breakdown: Movies vs. TV Shows
We examine the split between feature-length films and episodic series.

```python
type_counts = df['type'].value_counts()
print("Count by type:")
print(type_counts)
print("\nPercentage split:")
print((type_counts / len(df) * 100).round(2))
```

## 2. Release Year Distribution
Investigating the vintage of titles available on the platform.

```python
print("Oldest 5 titles in catalog:")
print(df.sort_values(by='release_year')[['title', 'type', 'release_year', 'country']].head(5))
print("\nTitles released per decade:")
df['decade'] = (df['release_year'] // 10) * 10
print(df['decade'].value_counts().sort_index())
```

## 3. Top Content Producing Nations
Handling co-productions by exploding comma-separated country fields.

```python
countries = df['country'].dropna().str.split(',').explode().str.strip()
countries = countries[countries != '']
top_10_countries = countries.value_counts().head(10)
print("Top 10 Content Producing Countries:")
print(top_10_countries)
```

## 4. Genre Popularity & Categorical Breakdown
Analyzing genres and listed_in categories across all titles.

```python
genres = df['listed_in'].dropna().str.split(',').explode().str.strip()
genres = genres[genres != '']
top_10_genres = genres.value_counts().head(10)
print("Top 10 Genres:")
print(top_10_genres)
```

## 5. Content Maturity Ratings
Analyzing the intended audience demographics for Netflix catalog.

```python
ratings = df['rating'].fillna('Unrated').value_counts()
print("Ratings Distribution:")
print(ratings)
```

## 6. Runtime and Season Length Distributions
Investigating movie duration in minutes vs TV show seasons.

```python
movie_runtimes = df[df['type'] == 'Movie']['duration_minutes'].dropna()
print("Movie Duration Summary (Minutes):")
print(movie_runtimes.describe())

tv_seasons = df[df['type'] == 'TV Show']['seasons_count'].dropna()
print("\nTV Show Seasons Summary:")
print(tv_seasons.value_counts().sort_index())
```

## 7. Addition Timeline: Ingestion Velocity
Examining when titles were ingested onto Netflix platform.

```python
additions = df['year_added'].dropna().astype(int).value_counts().sort_index()
print("Titles added per year:")
print(additions)
```
